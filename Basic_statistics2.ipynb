```python
import pandas as pd
import numpy as np
```

```python
patient_df = pd.read_csv("Patient_Data.csv")
billing_df = pd.read_csv("Billing_Data.csv")

patient_df.head()
```

```python
patient_df.info()
```

```python
patient_df = patient_df[['PatientID', 'Department', 'Doctor', 'BillAmount']]
patient_df.head()
```

```python
patient_df = patient_df.drop(columns=['ReceptionistID', 'CheckInTime'], errors='ignore')
```

```python
dept_bill = patient_df.groupby('Department')['BillAmount'].sum()
print(dept_bill)
```

```python
patient_df = patient_df.drop_duplicates(subset='PatientID')
```

```python
patient_df['BillAmount'].fillna(patient_df['BillAmount'].mean(), inplace=True)
```

```python
merged_df = pd.merge(patient_df, billing_df, on='PatientID', how='inner')
merged_df.head()
```

```python
new_patients = pd.DataFrame({
    'PatientID': [9991, 9992],
    'Department': ['Cardiology', 'Neurology'],
    'Doctor': ['Dr.A', 'Dr.B'],
    'BillAmount': [5000, 7000]
})

merged_df = pd.concat([merged_df, new_patients], ignore_index=True)
```

```python
merged_df['InsuranceCovered'] = [True]*len(merged_df)
merged_df['FinalAmount'] = merged_df['BillAmount'] * 0.9
```
