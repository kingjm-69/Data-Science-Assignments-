```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("sales_data_with_discounts.csv")
df.head()
```

## Descriptive Analytics

```python
num_cols = df.select_dtypes(include=np.number).columns
print(num_cols)
```

```python
for col in num_cols:
    print("Column:", col)
    print("Mean:", df[col].mean())
    print("Median:", df[col].median())
    print("Mode:", df[col].mode()[0])
    print("Std Dev:", df[col].std())
```

Mean = average value  
Median = middle value  
Mode = most frequent value  
Standard deviation = spread of data  

```python
df[num_cols].hist(figsize=(10,8))
plt.show()
```

```python
for col in num_cols:
    sns.boxplot(x=df[col])
    plt.title(col)
    plt.show()
```

```python
cat_cols = df.select_dtypes(include='object').columns
print(cat_cols)
```

```python
for col in cat_cols:
    df[col].value_counts().plot(kind='bar')
    plt.title(col)
    plt.show()
```

Bar charts show frequency of categories.

```python
print(df.isnull().sum())
```

```python
df.fillna(df.mean(numeric_only=True), inplace=True)
```

```python
df.drop_duplicates(inplace=True)
```

The dataset was analyzed using statistical methods and visualizations.
Histograms and boxplots helped identify patterns and outliers.
Bar charts showed category distribution.
Data preprocessing improved data quality.
