```python
import numpy as np
import scipy.stats as stats
```

```python
data = [1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85, 1.07, 1.48,
        1.20, 1.33, 1.18, 1.22, 1.29]

n = len(data)
mean = np.mean(data)
std = np.std(data, ddof=1)

print("Sample Mean:", mean)
print("Sample Std Dev:", std)
```

```python
confidence = 0.99
t_value = stats.t.ppf((1 + confidence)/2, df=n-1)

margin_error = t_value * (std / np.sqrt(n))

lower = mean - margin_error
upper = mean + margin_error

print("99% Confidence Interval (t-distribution):")
print(lower, "to", upper)
```

Since sample size is small (n < 30) and population standard deviation is unknown,
we use t-distribution.

```python
sigma = 0.2

z_value = stats.norm.ppf((1 + confidence)/2)

margin_error = z_value * (sigma / np.sqrt(n))

lower = mean - margin_error
upper = mean + margin_error

print("99% Confidence Interval (Z-distribution):")
print(lower, "to", upper)
```

###Hypothesis Testing

```python
x_bar = 3050   # sample mean
X = 600

mu = 1000 + 5*X   # theoretical mean
sigma = 5 * 25    # std deviation
n = 25

print("Theoretical Mean:", mu)
```

H0 (Null Hypothesis): μ = theoretical cost
H1 (Alternative Hypothesis): μ > theoretical cost

```python
z = (x_bar - mu) / (sigma / np.sqrt(n))
print("Z value:", z)
```

```python
alpha = 0.05
z_critical = stats.norm.ppf(1 - alpha)

print("Critical Value:", z_critical)
```

```python
if z > z_critical:
    print("Reject Null Hypothesis")
else:
    print("Fail to Reject Null Hypothesis")
```

Since the calculated Z value is greater than the critical value,
we reject the null hypothesis.

There is strong evidence that the weekly operating cost is higher than the model.
